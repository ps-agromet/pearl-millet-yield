```python
import os
import matplotlib.pyplot as plt
import numpy as np
from shapely.geometry import box
import geopandas as gpd
```

```python
import seaborn as sns
```

```python
from mpl_toolkits.axes_grid1 import make_axes_locatable
```

```python
# repository root (notebooks live in <repo>/notebooks)
base_loc = os.path.abspath(os.path.join(os.getcwd(), '..'))
```

```python
shpCV=gpd.read_file(os.path.join(base_loc,'data','location data','Rajasthan_PM_CV.shp'))
shpmean=gpd.read_file(os.path.join(base_loc,'data','location data','Rajasthan_PM_mean.shp'))
shpwet=gpd.read_file(os.path.join(base_loc,'data','location data','Rajasthan_PM_weather.shp'))
```

```python
shpwet['MinT']=shpwet['MinT'].astype(float)
shpwet['MaxT']=shpwet['MaxT'].astype(float)
shpwet['ET']=shpwet['ET'].astype(float)
shpwet['Rainfall']=round(shpwet['Rainfall'].astype(float),2)
```

```python
dates=['June 01', 'June 16', 'July 01', 'July 16',
       'July 31', 'August 15', 'August 30']
```

```python
def map_gen_yield(date):
    fig, ax = plt.subplots(1, 1)
    plt.title('Mean Yield(t/ha)'+'\n'+date)
    fn=os.path.join(base_loc,'maps','Mean-'+date+'.png')
    shpmean.plot(column=date,  scheme='quantiles',ax=ax, legend=True)
#     plt.legend(bbox_to_anchor=(1.09, 1), loc='upper left', borderaxespad=0)
    fig.savefig(fn,dpi=300,bbox_inches='tight')
```

```python
def map_gen_cv(date):
    fig, ax = plt.subplots(1, 1)
    plt.title('CV'+'\n'+date)
    fn=os.path.join(base_loc,'maps','CV-'+date+'.png')
    shpCV.plot(column=date,  scheme='quantiles',ax=ax, legend=True)
#     plt.legend(bbox_to_anchor=(1.09, 1), loc='upper left', borderaxespad=0)
    fig.savefig(fn,dpi=300,bbox_inches='tight')
```

```python
def map_gen_weather(date):
    fig, ax = plt.subplots(1, 1)
    plt.title('Mean'+' '+date)
    fn=os.path.join(base_loc,'maps','WeatherParam-'+date+'.png')
    shpwet.plot(column=date,  scheme='quantiles',  ax=ax, legend=True)
#     plt.legend(bbox_to_anchor=(1.09, 1), loc='upper left', borderaxespad=0)
    fig.savefig(fn,dpi=300,bbox_inches='tight')
```

```python
for d in dates:
    map_gen_yield(d)
    map_gen_cv(d)
```

```python
params=['MinT','MaxT','ET','Rainfall']
for d in params:
    map_gen_weather(d)
```
